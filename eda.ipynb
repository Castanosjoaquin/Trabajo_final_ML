```python
import pandas as pd

panel  = pd.read_parquet("data/processed/panel_nucleo.parquet")
magyp  = pd.read_parquet("data/processed/fuente_magyp.parquet")
oni    = pd.read_parquet("data/processed/fuente_oni.parquet")
nasa   = pd.read_parquet("data/processed/fuente_nasa_power.parquet")
ndvi   = pd.read_parquet("data/processed/fuente_ndvi.parquet")
```

```python
def df_info(name, df):
    print(f"{'='*60}")
    print(f"  {name}")
    print(f"{'='*60}")
    print(f"Shape        : {df.shape[0]:,} filas x {df.shape[1]} columnas")
    print(f"Memoria      : {df.memory_usage(deep=True).sum() / 1e6:.2f} MB")
    print(f"\n--- Tipos de datos ---")
    print(df.dtypes.value_counts().to_string())
    nans = df.isnull().sum()
    nans_pct = (nans / len(df) * 100).round(2)
    nan_df = pd.DataFrame({"nans": nans, "pct": nans_pct}).query("nans > 0").sort_values("pct", ascending=False)
    print(f"\n--- NaNs ({len(nan_df)} columnas con faltantes) ---")
    if len(nan_df):
        print(nan_df.to_string())
    else:
        print("Sin valores faltantes")
    print(f"\n--- Primeras filas ---")
    display(df.head(3))
    print(f"\n--- Describe (numéricas) ---")
    display(df.describe().T)
```

# Análisis exploratorio por fuente

## 1. Panel núcleo (`panel_nucleo`)

```python
df_info("panel_nucleo", panel)
```

## 2. Fuente MAGYP (`fuente_magyp`)

```python
df_info("fuente_magyp", magyp)
```

## 3. Fuente ONI (`fuente_oni`)

```python
df_info("fuente_oni", oni)
```

## 4. Fuente NASA Power (`fuente_nasa_power`)

```python
df_info("fuente_nasa_power", nasa)
```

## 5. Fuente NDVI (`fuente_ndvi`)

```python
df_info("fuente_ndvi", ndvi)
```
